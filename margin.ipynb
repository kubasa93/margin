```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# List to store data
data = []

# Create input widgets
qty_input = widgets.Textarea(
    value='',
    placeholder='Enter quantities, one per line, e.g.:\n5\n3\n2',
    description='Quantities:',
    layout=widgets.Layout(width='300px', height='150px')
)
cost_input = widgets.Textarea(
    value='',
    placeholder='Enter costs, one per line, e.g.:\n10.0\n15.5\n20.0',
    description='Costs:',
    layout=widgets.Layout(width='300px', height='150px')
)
margin_input = widgets.FloatText(value=20.0, description='Margin (%):')
submit_button = widgets.Button(description='Calculate Prices')
output = widgets.Output()

# Function to process inputs and update display
def calculate_prices(b):
    data.clear()  # Clear previous data
    margin = margin_input.value / 100  # Convert to decimal
    qty_lines = qty_input.value.strip().split('\n')
    cost_lines = cost_input.value.strip().split('\n')

    # Ensure same number of lines
    if len(qty_lines) != len(cost_lines):
        with output:
            clear_output(wait=True)
            print("Error: Number of quantities and costs must match.")
        return

    for qty_str, cost_str in zip(qty_lines, cost_lines):
        if qty_str.strip() and cost_str.strip():
            try:
                qty = float(qty_str)
                cost = float(cost_str)
                if margin >= 1:  # Prevent division by zero or negative
                    selling_price = 0.0
                else:
                    selling_price = cost / (1 - margin)
                data.append({'Quantity': int(qty), 'Cost per Unit': cost, 'Selling Price per Unit': selling_price})
            except ValueError:
                with output:
                    print(f"Invalid entry: Qty '{qty_str}' or Cost '{cost_str}' - Skipping")

    # Display results
    with output:
        clear_output(wait=True)
        if data:
            df = pd.DataFrame(data)
            # Calculate totals
            df['Total Cost'] = df['Quantity'] * df['Cost per Unit']
            df['Total Sales'] = df['Quantity'] * df['Selling Price per Unit']

            # Display the DataFrame
            display(df)

            # Display grand totals
            total_cost = df['Total Cost'].sum()
            total_sales = df['Total Sales'].sum()
            print(f"\nGrand Total Cost: ${total_cost:.2f}")
            print(f"Grand Total Sales: ${total_sales:.2f}")
        else:
            print("No valid items added.")

# Link button to function
submit_button.on_click(calculate_prices)

# Display the UI horizontally
input_box = widgets.HBox([qty_input, cost_input])
display(input_box, margin_input, submit_button, output)

```
